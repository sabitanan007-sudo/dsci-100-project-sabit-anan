#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)

womens_tidy <- womens_march |>
  # we keep  relevant columns
  select(seed, reg_wins_pct, first_game_at_home, tourney_finish) |>  #these ensure we meet the requirement as well have the data we need for our own analysis
  # next we need to filter
  filter(seed > 0,
         !is.na(seed),
         !is.na(reg_wins_pct),
         !is.na(first_game_at_home),
         !is.na(tourney_finish)) |>
  mutate(sweet_16 = as_factor(!(tourney_finish %in% c("opening_round_loss",   # TRUE if team reached the Sweet Sixteen
                                                      "first_round_loss",
                                                      "second_round_loss"))),
         # convert home variable to a factor
  first_game_at_home = as_factor(first_game_at_home)) |>
  # remove column used to build the response
  select(-tourney_finish)

head(womens_tidy)
```

```r
#let's check how many teams actually reached s16 and how many played first game at home

womens_tidy |>
  group_by(sweet_16) |>
  summarize(count = n())

womens_tidy |>
  group_by(first_game_at_home) |>
  summarize(count = n())
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
summary_stats <- womens_tidy |>
  # we keep numeric columns
  select(seed, reg_wins_pct) |>
  pivot_longer(everything(), names_to = "variable", values_to = "value") |>
  # compute stats separately for each variable
  group_by(variable) |>
  summarize(mean = mean(value, na.rm = TRUE),
            min = min(value, na.rm = TRUE),
            max = max(value, na.rm = TRUE))

summary_stats
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
missing_counts <- womens_tidy |>
 summarize(seed = sum(is.na(seed)),
          reg_wins_pct = sum(is.na(reg_wins_pct)),
          first_game_at_home = sum(is.na(first_game_at_home)),
          sweet_16 = sum(is.na(sweet_16)))|>
#

  # reshape into a readable two-column table
  pivot_longer(everything(), names_to = "column", values_to = "value_missing")

missing_counts
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.3: Summarize over levels of a categorical variable (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates mean of one numeric variable for each of the levels of another categorical variable. 
- Show the outputted table. 
</div>

```r
mean_by_home <- womens_tidy |>
  # split the data into groups: played at home (Y) vs. not (N)
  group_by(first_game_at_home) |>
  # average regular season win percentage within each group
  summarize(mean_reg_wins_pct = mean(reg_wins_pct))

mean_by_home
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

- Use `ggplot` to plot the distribution of one of the numeric variables in your tidied data set.

- Use `ggplot` to create a visualization showing the relative proportions of each category/level for one of categorical variable in your tidied data set. 

- Clearly label the variable that is being plotted in each plot.

*Notes:* 
- *Be sure the output is as clear as possible, provide an English title along with readable x and y column names*

- *You may also add more code cells to this section*

</div>

```r
ggplot(womens_tidy, aes(x = reg_wins_pct)) +
  # bars count how many teams fall in each 5-point bin
  geom_histogram(binwidth = 5, fill = "steelblue", color = "white") +
  labs(title = "Distribution of Regular Season Win Percentage",
       x = "Regular Season Win Percentage (%)",
       y = "Number of Teams") +
  theme(text = element_text(size = 14))
```

```r
womens_tidy |>
  # count teams in each level of the response
  group_by(sweet_16) |>
  summarize(count = n()) |>
  # convert counts to proportions
  mutate(prop = count / sum(count)) |>
  ggplot(aes(x = sweet_16, y = prop)) +
  geom_bar(stat = "identity", fill = "darkorange") +
  labs(title = "Proportion of Teams That Reached the Sweet Sixteen",
       x = "Reached the Sweet Sixteen?",
       y = "Proportion of Teams") +
  theme(text = element_text(size = 14))
```

Visualization 1: Home court vs. reaching the Sweet Sixteen

```r
womens_tidy |>
  # count teams for each home/outcome combination
  group_by(first_game_at_home, sweet_16) |>
  summarize(count = n()) |>
  # proportion within each home group (summarize drops the last group,
  # so the data is still grouped by first_game_at_home)
  mutate(prop = count / sum(count)) |>
  ggplot(aes(x = first_game_at_home, y = prop, fill = sweet_16)) +
  geom_bar(stat = "identity", position = "dodge") +
  labs(title = "Sweet Sixteen Outcome by Home Court in First Game",
       x = "Played First Game at Home?",
       y = "Proportion of Teams",
       fill = "Reached Sweet Sixteen?") +
  theme(text = element_text(size = 14))
```

Visualization 2: : Seed vs. regular season win percentage

```r
ggplot(womens_tidy, aes(x = seed, y = reg_wins_pct, colour = sweet_16)) +
  # transparency helps where many points overlap
  geom_point(alpha = 0.5) +
  labs(title = "Seed vs. Regular Season Win Percentage",
       x = "Seed (1 = best)",
       y = "Regular Season Win Percentage (%)",
       colour = "Reached Sweet Sixteen?") +
  theme(text = element_text(size = 14))
```
